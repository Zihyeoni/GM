#**차원 축소 (Dimensionality Reduction)**

고차원 데이터(즉, 많은 변수 또는 특징을 가진 데이터)를 더 적은 수의 차원으로 변환하는 기법  
이는 데이터의 중요한 정보를 유지하면서도 불필요한 노이즈나 상관성이 높은 변수를 제거하여 분석을 더 효율적으로 수행할 수 있게 함

고차원 데이터는 분석하기 어려울 수 있으며, 특히 시각화나 모델링 과정에서 과적합(overfitting)이나 계산 복잡성을 초래할 수 있음  
차원 축소는 이러한 문제를 해결하고 데이터의 핵심 구조를 이해하는 데 도움을 줌

image.png image.png

### 주요 차원 축소 기법

1. **주성분 분석(PCA, Principal Component Analysis)**: **가장 널리 사용되는 차원 축소 기법** 중 하나  
PCA는 데이터의 분산을 최대화하는 방향으로 새로운 축(주성분)을 생성하여, 이 축을 따라 데이터를 투영함으로써 차원을 축소
2. **선형 판별 분석(LDA, Linear Discriminant Analysis)**: 주로 분류 문제에서 사용되며, 서로 다른 클래스 간의 분리를 최대화하는 축을 찾음
3. **t-SNE (t-Distributed Stochastic Neighbor Embedding)**: 주로 시각화를 목적으로 사용되는 비선형 차원 축소 기법  
고차원 공간의 데이터를 저차원 공간(주로 2차원 또는 3차원)으로 변환하여 유사한 데이터 포인트들이 가까이 위치하도록 함
4. **UMAP (Uniform Manifold Approximation and Projection)**: t-SNE와 유사하지만 더 빠르고 시각적으로 해석 가능한 결과를 제공하는 비선형 차원 축소 기법

### 차원 축소의 응용 분야

1. **데이터 시각화**: 차원 축소를 통해 고차원 데이터를 2D 또는 3D로 축소하여 시각화할 수 있음  
이는 데이터의 패턴이나 클러스터를 시각적으로 탐색하는 데 유용
2. **노이즈 제거**: 차원 축소를 통해 데이터에서 노이즈를 제거하고 중요한 특징만을 남겨 분석 성능을 개선
3. **컴퓨터 비전**: 이미지 데이터의 차원을 축소하여 분석하거나, 고차원 피처를 줄여 처리 속도를 향상
4. **자연어 처리**: 텍스트 데이터의 차원을 축소하여 텍스트 간의 유사성을 비교하거나, 주제를 분석하는 데 사용

### 1. 공분산 행렬 계산

먼저, 데이터 매트릭스  $X$ 가 주어졌을 때, 각 특징의 평균을 빼서 중심화(centering)  
그 다음, 공분산 행렬 $\Sigma$를 계산:

$$
\Sigma = \frac{1}{n-1} X^T X
$$

여기서  $X$ 는 중심화된 데이터 행렬

- $\Sigma$는 **공분산 행렬**을 나타냄
- $n$은 데이터 포인트의 개수
- $X$는 데이터 행렬입니다. $X$는 각 행이 관측된 데이터 포인트를 나타내고, 각 열이 다른 변수(특징)를 나타냄  
$X$의 크기는 $n×p$이며, 여기서 $p$는 변수의 개수
- $X^T$는 데이터 행렬 $X$의 전치 행렬  
$X^T$의 크기는 $p×n$입니다.
- $X^TX$는 $p×p$ 크기의 행렬로, 이는 각 변수 간의 내적을 계산하여 변수 간의 관계를 나타냄

공분산 행렬 계산 과정

1. **데이터 행렬 $X: X$**는 표본 데이터 행렬로, $X_{ij}$는 $i$번째 데이터 포인트의 $j$번째 변수의 값을 나타냄
2. **데이터 중심화** : 공분산 행렬을 구하기 전에 보통 데이터 행렬 $X$를 중심화

 즉, 각 변수의 평균을 빼서 데이터의 중심을 원점에 맞춤  
 중심화된 데이터 행렬은 $X$를 변수의 평균을 기준으로 이동시킨 행렬
3. **전치 행렬 $X^T$**: 데이터 행렬 $X$의 전치 행렬 $X^T$는 행과 열을 바꾼 행렬
4. **행렬 곱 $X^T X: X^T$**와 $X$의 행렬 곱은 각 변수 쌍의 내적을 구하는 과정으로, 이는 변수 간의 공분산을 포함
5. **스케일링**: 결과 행렬을 $n−1$로 나누어 각 변수의 공분산을 얻음  
이때 $n−1$로 나누는 것은 샘플 공분산의 불편 추정량을 구하기 위함임

공분산 행렬의 의미

공분산 행렬 $\Sigma$는 $p \times p$ 크기의 행렬로, 다음과 같은 특징을 가짐:

- 행렬의 대각 원소는 각 변수의 **분산**을 나타냄
- 행렬의 비대각 원소는 변수 간의 **공분산**을 나타냄

### 2. 고유벡터와 고유값 계산

공분산 행렬 $\Sigma$의 고유벡터(eigenvector)와 고유값(eigenvalue)을 계산  
이 고유벡터가 주성분을 나타내며, 고유값은 이 주성분에 해당하는 분산의 크기를 나타냄

### 3. 주성분 선택

가장 큰 고유값에 해당하는 고유벡터를 선택하여, 데이터를 이 고유벡터들로 이루어진 새로운 축으로 투영

$$
Z = XW
$$

여기서  $W$ 는 선택된 주성분으로 이루어진 행렬,  $Z$ 는 축소된 차원 공간에서의 데이터

### **차원 축소의 장점**

- 데이터의 이해와 해석을 용이하게 함
- 시각화를 통해 고차원 데이터의 패턴을 파악 가능
- 모델의 계산 복잡도를 줄이고, 과적합을 방지 가능

다음은 차원 축소와 주성분 분석(PCA, Principal Component Analysis)을 이용한 데이터 압축 방법에 대한 예시입니다.

````python
!pip install scikit-learn opencv-python matplotlib

import cv2
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA

# 1. 이미지 로드 및 전처리
def load_image(image_path, target_size=(64, 64)):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)  # 이미지 흑백 변환
    img_resized = cv2.resize(img, target_size)  # 이미지 크기 조정
    return img_resized

# 2. 이미지 차원 축소 및 복원
def reduce_and_reconstruct_image(image, n_components=1):
    pca = PCA(n_components=n_components)
    original_shape = image.shape
    image_flattened = image.flatten().reshape(1, -1)  # 이미지를 1차원 벡터로 변환

    # PCA 차원 축소
    image_reduced = pca.fit_transform(image_flattened)

    # 복원
    image_reconstructed = pca.inverse_transform(image_reduced)
    image_reconstructed = image_reconstructed.reshape(original_shape)  # 원래 이미지 형태로 복원

    return image_reconstructed

# 3. 이미지 저장 및 시각화
def save_and_display_images(original_image, reconstructed_image, output_path='reconstructed_image.png'):
    # 이미지 저장
    cv2.imwrite(output_path, reconstructed_image)

    # 시각화
    plt.figure(figsize=(8, 4))
    plt.subplot(1, 2, 1)
    plt.imshow(original_image, cmap='gray')
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(reconstructed_image, cmap='gray')
    plt.title('Reconstructed Image')
    plt.axis('off')

    plt.show()

# 4. 메인 실행 함수
def main(image_path, output_path='reconstructed_image.png', n_components=1):
    original_image = load_image(image_path)
    reconstructed_image = reduce_and_reconstruct_image(original_image, n_components=n_components)
    save_and_display_images(original_image, reconstructed_image, output_path)

# 예제 실행
image_path = '1.png'  # 입력 이미지 경로를 설정합니다.
main(image_path)
````

### **그로스 마케팅에서 차원 축소 활용 예제**

**예제 상황**

- **목표:** 고객 데이터를 차원 축소하여 그룹별 특성 분석
- **데이터:** 웹 방문 횟수, 광고 클릭 수, 구매 금액 등
- **기법:** PCA(주성분 분석) 활용

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 하드 코딩된 고객 행동 데이터 (그로스 마케팅)
data = {
    "Page_Visits": [15, 25, 30, 80, 95, 120, 150, 200, 250, 300,
                    320, 400, 450, 500, 550, 600, 700, 750, 800, 850],
    "Ad_Clicks": [1, 3, 5, 10, 12, 20, 25, 35, 40, 50,
                  60, 70, 80, 90, 100, 110, 120, 130, 140, 150],
    "Purchases": [0, 1, 1, 3, 5, 6, 10, 12, 15, 18,
                  20, 25, 30, 35, 40, 45, 50, 55, 60, 65],
    "Revenue": [50, 80, 120, 200, 250, 300, 500, 700, 900, 1100,
                1300, 1600, 2000, 2400, 2800, 3200, 3700, 4200, 4700, 5200]
}

# 데이터프레임 변환
df = pd.DataFrame(data)

# 데이터 정규화 (PCA를 위해 필수)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# PCA 적용 (2차원으로 축소)
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

# 변환된 결과를 데이터프레임으로 변환
df_pca = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])

# K-means를 이용한 고객 군집화 (3개 그룹)
kmeans = KMeans(n_clusters=3, random_state=42)
df_pca['Cluster'] = kmeans.fit_predict(df_pca)

# 차원 축소된 데이터 시각화
plt.figure(figsize=(8, 6))
plt.scatter(df_pca['PC1'], df_pca['PC2'], c=df_pca['Cluster'], cmap='viridis', alpha=0.7)
plt.xlabel("주성분 1 (PC1)")
plt.ylabel("주성분 2 (PC2)")
plt.title("PCA 기반 고객 행동 데이터 차원 축소 및 군집화")
plt.colorbar(label="Cluster")
plt.show()
````

- **차원 축소 필요성:** 원래 4차원 데이터(페이지 방문, 광고 클릭, 구매, 매출) → 2차원으로 축소하여 시각화 가능
- **군집화(K-means):** 차원 축소 후 유사한 고객을 3개 그룹으로 분류
- **활용 사례:** 특정 그룹별 맞춤형 마케팅 전략 적용 가능

## **계층적 군집 분석 (Hierarchical Clustering)**

### (1) **병합형(Agglomerative) 계층적 군집 분석**

- **하향식(Top-Down) 방식**이 아니라 **상향식(Bottom-Up) 방식**으로 진행
- 개별 데이터 포인트를 하나의 군집으로 간주하고, 유사한 군집끼리 반복적으로 병합하여 최종적으로 하나의 큰 군집이 될 때까지 수행
- 일반적인 단계:
    1. 모든 데이터를 개별 군집으로 설정
    2. 두 개의 가장 가까운 군집을 병합
    3. 위 과정을 반복하여 트리를 형성
    4. 최종적으로 하나의 군집이 형성되면 종료

### (2) **분할형(Divisive) 계층적 군집 분석**

- 병합형과 반대로 **하향식(Top-Down) 방식**으로 진행
- 모든 데이터를 하나의 큰 군집으로 시작한 후, 점진적으로 가장 비슷하지 않은 데이터 포인트를 분할하여 더 작은 군집을 형성
- 병합형보다 연산량이 크고 덜 자주 사용됨

### **군집 간 거리 측정 방법 (Linkage Methods)**

군집을 병합할 때, 군집 간 거리를 정의하는 방법 필요

| Linkage 방법 | 설명 |
| --- | --- |
| **단일 연결법 (Single Linkage)** | 가장 가까운 두 점 간 거리 |
| **완전 연결법 (Complete Linkage)** | 가장 먼 두 점 간 거리 |
| **평균 연결법 (Average Linkage)** | 모든 점 사이 거리의 평균 |
| **중심 연결법 (Centroid Linkage)** | 군집의 중심(평균) 간 거리 |
| **워드 방법 (Ward's Method)** | 군집 내 분산 증가 최소화 |

### **덴드로그램(Dendrogram)**

- 계층적 군집 분석의 결과를 **트리 구조**로 표현한 그래프
- Y축은 거리 또는 유사도를 나타냄
- 특정 높이에서 자르면 해당 높이에 해당하는 군집 개수 결정 가능

### **장점과 단점**

**장점**

- 군집 개수를 미리 지정할 필요 없음
- 덴드로그램을 통해 군집의 관계를 시각적으로 분석 가능
- 병합형 방법은 데이터의 밀도와 관계없이 사용 가능

**단점**

- 데이터가 많을 경우 연산량이 매우 커짐 (시간 복잡도: O(n²) 또는 O(n³))
- 이상치(Outlier)에 영향을 많이 받음
- 병합형 방식은 한 번 합쳐진 군집이 다시 분리되지 않음

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

# 샘플 데이터 생성
np.random.seed(42)
X = np.random.rand(10, 2)

# 계층적 군집 분석 수행
linked = linkage(X, method='ward')

# 덴드로그램 시각화
plt.figure(figsize=(10, 5))
dendrogram(linked, labels=range(1, 11), leaf_rotation=90)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
````

### **시나리오: 고객 세분화를 통한 맞춤형 마케팅 전략 수립**

기업 A는 온라인 쇼핑몰을 운영하고 있으며, 고객별로 마케팅 전략을 최적화하려 합니다. 이를 위해 계층적 군집 분석을 이용해 고객을 몇 개의 그룹으로 분류하고, 각 그룹에 맞는 마케팅 전략을 적용하려 합니다.

분석할 고객 데이터에는 다음과 같은 변수가 포함됩니다.

1. **월 평균 구매 금액** (`Monthly_Spending`)
2. **방문 빈도** (`Visit_Frequency`)
3. **고객 평점** (`Customer_Satisfaction`)

목표:

- 고객을 **유사한 소비 행동**을 보이는 그룹으로 분류
- 각 그룹별 맞춤형 프로모션 전략 수립
- 예를 들어, VIP 고객에게는 **할인 쿠폰** 제공, 방문 빈도가 낮은 고객에게는 **재방문 유도 캠페인** 적용


아래 코드는 **샘플 고객 데이터**를 생성하고 계층적 군집 분석을 수행한 후 **덴드로그램과 클러스터링 결과**를 시각화합니다.

````python
pip install ace_tools_open
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster

# 1. 샘플 고객 데이터 생성 (월 평균 구매 금액, 방문 빈도, 고객 평점)
np.random.seed(42)
num_customers = 20

customer_data = pd.DataFrame({
    'Customer_ID': range(1, num_customers + 1),
    'Monthly_Spending': np.random.randint(50, 500, size=num_customers),   # 월 평균 구매 금액 ($)
    'Visit_Frequency': np.random.randint(1, 20, size=num_customers),      # 월 방문 빈도 (회)
    'Customer_Satisfaction': np.random.randint(1, 10, size=num_customers) # 고객 평점 (1~10)
})

# 2. 계층적 군집 분석 수행
X = customer_data[['Monthly_Spending', 'Visit_Frequency', 'Customer_Satisfaction']].values
linked = linkage(X, method='ward')

# 3. 덴드로그램 시각화
plt.figure(figsize=(10, 5))
dendrogram(linked, labels=customer_data['Customer_ID'].values, leaf_rotation=90)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customer ID")
plt.ylabel("Distance")
plt.show()

# 4. 최적의 군집 개수를 지정하여 클러스터 할당 (예: 3개)
num_clusters = 3
customer_data['Cluster'] = fcluster(linked, num_clusters, criterion='maxclust')

# 5. 군집별 고객 분포 시각화
plt.figure(figsize=(8, 6))
sns.scatterplot(data=customer_data, x="Monthly_Spending", y="Visit_Frequency", hue="Cluster", palette="viridis", s=100)
plt.title("Customer Segmentation using Hierarchical Clustering")
plt.xlabel("Monthly Spending ($)")
plt.ylabel("Visit Frequency (times/month)")
plt.legend(title="Cluster")
plt.show()

# 6. 결과 데이터 출력
import ace_tools_open as tools
tools.display_dataframe_to_user(name="Customer Segmentation Data", dataframe=customer_data)
````

**분석 결과 및 활용 방안**

1. **군집 분석 결과 해석**
    - 군집 1: **고액 지출 고객** (VIP 고객) → **프리미엄 혜택, 멤버십 제안**
    - 군집 2: **방문 빈도는 높지만 소비는 적은 고객** → **적립금, 무료 배송 혜택 제공**
    - 군집 3: **방문 빈도와 소비 모두 낮은 고객** → **리마케팅, 타겟 광고 활용**
2. **활용 방법**
    - 각 군집별 맞춤형 마케팅 캠페인 기획
    - 이메일, 푸시 알림, 할인 쿠폰 등 개인화된 마케팅 전략 실행
    - 방문 빈도가 낮은 고객을 대상으로 특별 이벤트 기획하여 재방문 유도

##**회귀분석을 통한 시계열 데이터 예측**

회귀 분석(Regression Analysis)은 종속 변수와 독립 변수 간의 관계를 찾고, 이를 바탕으로 미래 값을 예측하는 기법  
시계열 데이터(Time Series Data)에서도 회귀 분석을 활용하여 트렌드 패턴을 분석하고 미래 값을 예측할 수 있음

### **1. 시계열 데이터 예측에 회귀 분석을 적용하는 방법**

1. **특징 엔지니어링 (Feature Engineering)**
    
    시계열 데이터에는 `시간(Year, Month, Day)`과 같은 변수를 독립 변수(특징)로 사용
    
    이동 평균, 지수 가중 이동 평균(Exponential Moving Average, EMA) 등을 추가하면 예측 정확도를 높일 수 있음
    
2. **회귀 모델 선택**
    - **선형 회귀(Linear Regression)**: 기본적인 추세 예측
    - **다항 회귀(Polynomial Regression)**: 비선형 추세 예측 가능
    - **랜덤 포레스트 회귀(Random Forest Regression)**: 시계열 변동성이 클 경우 강력한 성능을 발휘
    - **LSTM, XGBoost** 등 딥러닝/머신러닝 모델도 활용 가능하지만, 기본 회귀 분석도 강력한 도구가 될 수 있음

### **2. 예제: 회귀 분석을 이용한 시계열 데이터 예측**

**가정: 매출 데이터 예측**

기업이 지난 3년간 월별 매출 데이터를 보유하고 있으며, 이를 바탕으로 향후 6개월의 매출을 예측하려 한다.

**선형 회귀를 활용한 시계열 예측**

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# 1. 데이터 생성 (지난 3년간 월별 매출 데이터)
date_range = pd.date_range(start="2021-01-01", periods=36, freq="M")
np.random.seed(42)
sales = np.linspace(100, 300, num=36) + np.random.randint(-20, 20, size=36)  # 추세 + 변동성 추가

# 데이터프레임 구성
data = pd.DataFrame({'Date': date_range, 'Sales': sales})
data['Month'] = data['Date'].dt.month
data['Year'] = data['Date'].dt.year
data['Time_Index'] = range(1, len(data) + 1)  # 시간 순서 인덱스 생성

# 2. 독립 변수(X)와 종속 변수(y) 설정
X = data[['Time_Index']]
y = data['Sales']

# 3. 훈련 데이터와 테스트 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# 4. 선형 회귀 모델 학습
model = LinearRegression()
model.fit(X_train, y_train)

# 5. 향후 6개월 예측
future_dates = pd.date_range(start="2024-01-01", periods=6, freq="M")
future_index = np.arange(len(data) + 1, len(data) + 7).reshape(-1, 1)
future_sales = model.predict(future_index)

# 6. 시각화: 실제 vs 예측
plt.figure(figsize=(10, 5))
plt.plot(data['Date'], data['Sales'], label="Actual Sales", marker='o')
plt.plot(future_dates, future_sales, label="Predicted Sales", marker='s', linestyle="dashed")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Sales Prediction using Linear Regression")
plt.legend()
plt.xticks(rotation=45)
plt.show()
````

#### **3. 해석**

1. 데이터 생성
    
    3년간 월별 매출 데이터를 생성하고, `Time_Index`를 독립 변수로 사용하였다.
    
2. 모델 학습
    
    `Time_Index`를 기반으로 선형 회귀 모델을 훈련하였다.
    
3. 미래 예측
    
    향후 6개월(2024년 1월~6월)의 매출을 예측하고 시각화하였다.
    

#### **4. 결과 활용**

1. 매출 추세 분석 및 예측 기반 마케팅 전략 수립
2. 특정 이벤트 이후의 매출 증가/감소 예측
3. 공급망 최적화: 향후 매출 증가 대비 재고 조정


시계열 데이터에서 회귀 분석을 활용하면 비교적 간단한 방식으로 미래 값을 예측할 수 있다. 추세가 선형적일 경우 **선형 회귀**, 변동성이 클 경우 **랜덤 포레스트 회귀 또는 LSTM**을 활용하여 보다 정교한 예측이 가능하다.

### **회귀분석을 이용한 그로스마케팅**

아래는 **그로스 마케팅 성과 예측을 위한 데이터 생성, 이동 평균(MA) 및 지수 가중 이동 평균(EMA) 적용한 내용입니다.**

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 랜덤 시드 설정
np.random.seed(42)

# 날짜 데이터 생성 (100일치)
dates = pd.date_range(start="2024-01-01", periods=100, freq='D')

# 하드 코딩된 기본 지표 (그로스 마케팅 성과 관련 데이터)
page_visits = np.random.randint(50, 1000, 100)  # 페이지 방문 수
ad_clicks = np.random.randint(5, 200, 100)  # 광고 클릭 수
purchases = np.random.randint(0, 50, 100)  # 구매 수
revenue = purchases * np.random.randint(10, 500, 100)  # 매출 (구매 * 평균 상품 가격)
session_duration = np.random.randint(1, 120, 100)  # 세션 지속 시간 (분)
bounce_rate = np.random.uniform(10, 80, 100)  # 이탈률 (%)

# 데이터프레임 생성
df_growth = pd.DataFrame({
    "Date": dates,
    "Page_Visits": page_visits,
    "Ad_Clicks": ad_clicks,
    "Purchases": purchases,
    "Revenue": revenue,
    "Session_Duration": session_duration,
    "Bounce_Rate": bounce_rate
})

# 이동 평균 (7일 평균, 단기 트렌드)
df_growth["Revenue_MA7"] = df_growth["Revenue"].rolling(window=7).mean()

# 지수 가중 이동 평균 (EMA, 7일)
df_growth["Revenue_EMA7"] = df_growth["Revenue"].ewm(span=7, adjust=False).mean()

# 시각화
plt.figure(figsize=(12, 6))

# 원본 매출 데이터 시각화
plt.plot(df_growth["Date"], df_growth["Revenue"], label="Revenue", color="blue", alpha=0.5)

# 이동 평균 (MA) 시각화
plt.plot(df_growth["Date"], df_growth["Revenue_MA7"], label="7-day MA", color="red", linestyle="dashed")

# 지수 가중 이동 평균 (EMA) 시각화
plt.plot(df_growth["Date"], df_growth["Revenue_EMA7"], label="7-day EMA", color="green", linestyle="dotted")

# 제목 및 축 라벨 추가
plt.title("그로스 마케팅 성과 예측 - 이동 평균 및 EMA 비교")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True)

# 그래프 출력
plt.show()
````

#### **1. 데이터 생성**

- `Date`: 100일간의 날짜 생성
- `Page_Visits`: 웹사이트 방문 수 (50~1000 사이 랜덤 값)
- `Ad_Clicks`: 광고 클릭 수 (5~200 사이 랜덤 값)
- `Purchases`: 구매 횟수 (0~50 사이 랜덤 값)
- `Revenue`: 구매 횟수에 따라 랜덤 매출 생성 (구매 * 랜덤 상품 가격)
- `Session_Duration`: 방문 세션 지속 시간 (1~120분 사이 랜덤 값)
- `Bounce_Rate`: 이탈률 (10%~80% 사이 랜덤 값)


#### **2. 이동 평균 (MA) & 지수 가중 이동 평균 (EMA)**

- **이동 평균(MA)**
    - `Revenue_MA7`: 매출의 **7일 이동 평균**을 계산하여 단기 트렌드 확인
    - 최근 7일의 평균 매출을 계산하여 변동성을 줄이고 패턴을 분석할 수 있음
- **지수 가중 이동 평균(EMA)**
    - `Revenue_EMA7`: 최근 데이터에 더 높은 가중치를 부여하는 **7일 지수 가중 이동 평균(EMA)**
    - 최근 데이터일수록 가중치를 높게 부여하여 즉각적인 변화를 빠르게 반영할 수 있음


#### **3. 시각화**

- **파란색 실선**: 원본 매출 데이터
- **빨간색 점선**: 7일 이동 평균 (단기 트렌드 확인)
- **초록색 점선**: 7일 지수 가중 이동 평균 (최근 변화 반영)
- **그래프 설명**
    - `plt.plot()`을 사용해 원본 데이터, 이동 평균(MA), EMA를 각각 다른 스타일로 출력
    - `plt.legend()`로 그래프 범례 추가
    - `plt.grid(True)`로 가이드라인 표시

#### **활용 가능성**

- **그로스 마케팅 성과 분석** → 매출 데이터의 변동성을 줄여 트렌드 분석 가능
- **단기 및 장기 패턴 예측** → 이동 평균(MA)과 지수 가중 이동 평균(EMA) 비교를 통해 마케팅 전략 최적화
- **시계열 예측 모델 적용** → 머신러닝 모델과 결합하여 향후 매출 예측 가능
